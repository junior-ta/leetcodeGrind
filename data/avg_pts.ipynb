```python
#Using statistics, teams, and seasons, calculate the average points per game (PTS / G) for every team in each season.
```

```python
import pandas as pd

stats= pd.read_csv("teams_stats.csv")
teams= pd.read_csv("teams.csv")
seasons= pd.read_csv("seasons.csv")
```

```python
yrs= seasons[["Season_ID"]]
tms= teams[["Team_ID" , "Team_Abbreviation"]]

df= yrs.merge(tms, how='cross')
df= df.merge(stats[["Season_ID","Team_ID","PTS"]], on=["Season_ID","Team_ID"])
```
