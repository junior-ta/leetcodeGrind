```python
import pandas as pd

stats= pd.read_csv("statistics.csv")
teams= pd.read_csv("teams.csv")
seasons= pd.read_csv("seasons.csv")
```

```python
merged = stats.merge(
    teams,
    on="Team_ID"
).merge(
    seasons,
    on="Season_ID"
)

team_ppg = (
    merged.groupby(["Season_Years", "Team_ID"])
    .agg({
        "PTS": "sum",
        "G": "sum"
    })
    .reset_index()
)

team_ppg["Team_PPG"] = (
    team_ppg["PTS"] / team_ppg["G"]
).round(2)

team_ppg = team_ppg.sort_values(
    ["Season_Years", "Team_PPG"],
    ascending=[True, False]
)
```
